# Importance of Feature selection:
Feature selection is the process of choosing only the most useful input features for a machine learning model. It helps improve model performance, reduces noise and makes results easier to understand.

1.Helps remove irrelevant and redundant features
2.Improves accuracy and reduces overfitting
3.Speeds up model training
4.Makes models simpler and easier to interpret

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, chi2

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
dataset1=pd.read_csv("Heart_Dataset.csv.csv",index_col=None)
df2=dataset1
df2 = pd.get_dummies(df2, dtype=int, drop_first=0 or 1)
```

```python
indep_X=['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach',
       'exang', 'oldpeak', 'slope', 'ca', 'thal', ]
dep_Y=['target']
```

```python
dataset1=pd.read_csv("Heart_Dataset.csv.csv",index_col=None)

df2=dataset1

df2 = pd.get_dummies(df2, drop_first=True)

indep_X=df2.drop('target', axis=1)
dep_Y=df2['target']
```

```python

def split_scalar(indep_X, dep_Y):

    X_train, X_test, y_train, y_test = train_test_split(
        indep_X,
        dep_Y,
        test_size=0.25,
        random_state=0
    )

    scaler = StandardScaler()

    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    return X_train, X_test, y_train, y_test
```

```python

def selectkbest(indep_X, dep_Y, n):

    selector = SelectKBest(score_func=chi2, k=n)

    selected_features = selector.fit_transform(
        indep_X, dep_Y
    )

    return selected_features
```

```python

def cm_prediction(classifier, X_test, y_test):

    y_pred = classifier.predict(X_test)

    Accuracy = accuracy_score(y_test, y_pred)

    report = classification_report(
        y_test, y_pred, zero_division=0
    )

    cm = confusion_matrix(y_test, y_pred)

    return classifier, Accuracy, report, X_test, y_test, cm
```

```python

def logistic(X_train, y_train, X_test, y_test):

    classifier = LogisticRegression(random_state=0)
    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def svm_linear(X_train, y_train, X_test, y_test):

    classifier = SVC(kernel="linear", random_state=0)
    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def svm_NL(X_train, y_train, X_test, y_test):

    classifier = SVC(kernel="rbf", random_state=0)
    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def knn(X_train, y_train, X_test, y_test):

    classifier = KNeighborsClassifier(
        n_neighbors=7,
        metric="minkowski",
        p=2
    )

    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def Navie(X_train, y_train, X_test, y_test):

    classifier = GaussianNB()
    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def Decision(X_train, y_train, X_test, y_test):

    classifier = DecisionTreeClassifier(
        criterion="entropy",
        random_state=0
    )

    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def random(X_train, y_train, X_test, y_test):

    classifier = RandomForestClassifier(random_state=0)
    classifier.fit(X_train, y_train)

    return cm_prediction(classifier, X_test, y_test)
```

```python

def selectk_Classification(
    acclog, accsvml, accsvmnl,
    accknn, accnav, accdes, accr
):

    dataframe = pd.DataFrame(
        [[
            acclog[0],
            accsvml[0],
            accsvmnl[0],
            accknn[0],
            accnav[0],
            accdes[0],
            accr[0]
        ]],
        index=["ChiSquare"],
        columns=[
            "Logistic",
            "SVM Linear",
            "SVM Non-Linear",
            "KNN",
            "Naive Bayes",
            "Decision Tree",
            "Random Forest"
        ]
    )

    return dataframe
```

```python

# Check that input data is non-negative for Chi-square
if np.any(np.asarray(indep_X) < 0):
    raise ValueError(
        "Chi-square requires non-negative input features."
    )

# Select the best 6 features
kbest = selectkbest(indep_X, dep_Y, 6)

# Split into training and testing data
X_train, X_test, y_train, y_test = split_scalar(
    kbest, dep_Y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python

acclog = []
accsvml = []
accsvmnl = []
accknn = []
accnav = []
accdes = []
accrf = []
```

```python

# Logistic Regression
classifier, Accuracy, report, _, _, cm = logistic(
    X_train, y_train, X_test, y_test
)
acclog.append(Accuracy)
print("Logistic Regression:", Accuracy)


# Linear SVM
classifier, Accuracy, report, _, _, cm = svm_linear(
    X_train, y_train, X_test, y_test
)
accsvml.append(Accuracy)
print("Linear SVM:", Accuracy)


# Non-linear SVM
classifier, Accuracy, report, _, _, cm = svm_NL(
    X_train, y_train, X_test, y_test
)
accsvmnl.append(Accuracy)
print("Non-linear SVM:", Accuracy)


# K-Nearest Neighbors
classifier, Accuracy, report, _, _, cm = knn(
    X_train, y_train, X_test, y_test
)
accknn.append(Accuracy)
print("KNN:", Accuracy)


# Naive Bayes
classifier, Accuracy, report, _, _, cm = Navie(
    X_train, y_train, X_test, y_test
)
accnav.append(Accuracy)
print("Naive Bayes:", Accuracy)


# Decision Tree
classifier, Accuracy, report, _, _, cm = Decision(
    X_train, y_train, X_test, y_test
)
accdes.append(Accuracy)
print("Decision Tree:", Accuracy)


# Random Forest
classifier, Accuracy, report, _, _, cm = random(
    X_train, y_train, X_test, y_test
)
accrf.append(Accuracy)
print("Random Forest:", Accuracy)
```

```python

result = selectk_Classification(
    acclog,
    accsvml,
    accsvmnl,
    accknn,
    accnav,
    accdes,
    accrf
)

print("Classification Model Accuracy Comparison:")
display(result)
```

```python

# Select the Best 6 Features and Display Their Scores

from sklearn.feature_selection import SelectKBest, chi2
import pandas as pd

# Apply Chi-Square feature selection
selector = SelectKBest(score_func=chi2, k=6)
selector.fit(indep_X, dep_Y)

# Get all feature scores and selection status
feature_ranking = pd.DataFrame({
    "Feature": indep_X.columns,
    "ChiSquare Score": selector.scores_,
    "Selected": selector.get_support()
})

# Display only the selected 6 features, ranked by score
feature_ranking = (
    feature_ranking[feature_ranking["Selected"]]
    .sort_values("ChiSquare Score", ascending=False)
    .reset_index(drop=True)
)

feature_ranking.index = feature_ranking.index + 1
feature_ranking.index.name = "Rank"

print("Best 6 Selected Features:")
display(feature_ranking)
```
