```python
#Import Libraries


import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
#Prepare input and output
dataset=pd.read_csv("Heart_Dataset.csv.csv",index_col=None)
df2=dataset
dataset = pd.get_dummies(df2, dtype=int, drop_first=0 or 1)
# Selected independent features
features = ["thalach", "oldpeak", "ca", "cp", "exang", "chol"]

# Dependent feature
X = dataset[features]
y = dataset["target"]

print("Input columns:", features)
print("Output column: target")
print("Dataset shape:", X.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=0,
    stratify=y
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
import pickle
filename="finalized_model_LogisticRegression.sav"
```
